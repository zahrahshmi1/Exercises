```python
def avglengh(s):
    b=0
    m=0
    c=[]
    for _ in s:
        if ',' in s:
            s=s.replace(','," ")
        if ';' in s:
            s=s.replace(';'," ")
        if ':' in s:
            s=s.replace(':'," ")
        if '.' in s:
            s=s.replace('.'," ")
    a=s.split(' ')
    for _ in a:
        if '' in a:
            a.remove('')
    for i in a:
        b=len(i)
        c.append(b)
    for j in c:
        m+=j
    m=int(m/len(c))
    return(m)
```
