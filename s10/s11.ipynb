```python
class Weapon:
    def  __init__(self,name,damage,range,typ):
        self.name = name
        self.damage = damage
        self.range = range
        self.typ = typ

class Character:
    def __init__(self,name,health = 100 ,attack_power = 10 ,defens = 5):
        self.name = name
        self.health = health
        self.attack_power = attack_power
        self.defens = defens
        self.weapon = None
        self.experience = 0
        self.level = 1
        self.is_alive = True
    def increas_exp(self):
        self.experience += 1
    def increas_level(self):
        self.level += 1

    def choice_weapon(self,weapon):
        self.weapon = weapon

    def take_damage(self, damage):
        real_damage = damage - self.defens
        self.health -= real_damage
        if self.health <= 0:
            self.health = 0
            self.is_alive = False
            return (f'{self.name} died') 
    
    def attack(self,enemy):
        attack_damage = self.attack_power + self.weapon.damage - enemy.defens
        enemy.health -= attack_damage
        if enemy.health <= 0:
            enemy.health = 0
            enemy.is_alive = False
            return(f'{enemy} is deid')
        return attack_damage
    
class Soldier(Character):
    def __init__(self, name):
        super().__init__(name, health = 120 , attack_power = 15, defens = 8)
        self.choice_weapon(Weapon('AK-47',25,50,'gun'))

class Sniper(Character):
    def __init__(self, name):
        super().__init__(name, health = 80, attack_power = 12, defens = 4)
        self.choice_weapon(Weapon('AWM',55,200,'gun'))

    def attack(self,enemy):
         return super().attack(enemy)


class Medic(Character):
    def __init__(self, name):
        super().__init__(name, health = 100 , attack_power = 5, defens = 6)
    def heal(self,friend):

        if friend.is_alive:
            friend.health += 30
            return (f'{friend} health increas')

    
class Battle:
    def __init__(self,ch1,ch2):
        self.ch1 = ch1
        self.ch2 = ch2
    def fight(self):
        while self.ch2.is_alive and self.ch1.is_alive:
            self.ch1.attack(self.ch2)
        if self.ch1.is_alive:
            print ('ch1 win')
        else:
            print ('ch2 win')
        


class Mission:
    def __init__(self,title,level,enemy,goal):
        self.title = title
        self.level = level
        self.enemy = enemy
        self.goal = goal



class Level:
    def __init__(self,name):
        self.name = name
    def level_up(self):
        self.name.fight(Battle)
        if self.name.is_alive:
           self.increas_exp()
        if self.experience(Character) >= 10:
            self.increas_level
```
