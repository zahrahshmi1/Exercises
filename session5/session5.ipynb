```python
list_1=['milk','suger','butter','yogurt','cheese']
list_2=[5,2,10,1,3]
j=0
d = {}
for i in list_1:
    d[i]=list_2[j]
    j+=1 
a=[]
for t in d:
    a.append(d[t])
b=a[0]
c=a[0]
for n in a[1:]:
    if n>=b:
        b=n
    if n<c:
        c=n
print({f'max is {b} and min is {c}'})


    
```
